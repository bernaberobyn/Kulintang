```python
import pandas as pd

# load the main csv
df_main = pd.read_csv(MAIN_CSV)
df_main.head()
```

## Set the label column

```python
LABEL_COLUMN = 'label'
```

## Checking label distribution

```python
# Check the distribution of unique elements in Label  Column
df_main = df_main[df_main[LABEL_COLUMN] != 'n']
df_main = df_main[df_main[LABEL_COLUMN] != '9']
print(df_main[LABEL_COLUMN].value_counts())
```

## Split the dataset into train and test

```python
from sklearn.model_selection import train_test_split

# Split the main dataframe to train and test
df_train, df_test = train_test_split(df_main, test_size = 0.2, stratify=df_main[LABEL_COLUMN],random_state=42)
```

## Reset index of the training data

```python
# Reset the indexes
df_train.reset_index(drop=True)
```

## Check the label distribution of the training data

```python
# Check the distribution of unique elements in Label  Column
print(df_train[LABEL_COLUMN].value_counts())
```

## Reset the index of the test data

```python
df_test.reset_index(drop=True)
```

## Check the label distribution of the test data

```python
# Check the distribution of unique elements in Label  Columnp
print(df_test[LABEL_COLUMN].value_counts())
```

## Save Training and Test data to csv files

```python
# Save train and test sets to CSV files
df_train.to_csv('train_set.csv', index=False)
df_test.to_csv('test_set.csv', index=False)
```
